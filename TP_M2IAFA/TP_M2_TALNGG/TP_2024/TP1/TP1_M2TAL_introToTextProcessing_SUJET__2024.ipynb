# TP 1 : pre-processing texts

In this practical session, we will see a few basic processing of textual data.

Within a computer, text is encoded as a string of characters.
In order to analyze textual data within NLP applications, we first need to properly preprocess it.
An NLP preprocessing pipeline generally consists of the following steps :
* sentence segmentation
* tokenisation
* normalization: lower-casing, lemmatization, optionally removing stop-words and punctuation
* pos-tagging
* named entity recognition
* parsing

The first two steps are necessary, while the others are optional.

For these exercises, we will mainly use the module **spacy** (already installed on google colab, but some additional libraries might be missing).

Spacy provides ways to carry out tasks such as segmentation, tokenization, lemmatization and pos-tagging.

```python
import re

def get_nounchunk(doc,target_token):
    for chunk in doc.noun_chunks:
        if target_token.lower() in chunk.text.lower():
            return chunk.text
    return target_token        

def forme2_qui (doc):
    pattern = r'(de|pour|par)\s+qui\s*[?!]'
    match = re.search(pattern, doc.text)
    if match:
       return True
    return False   

def retrieve_nmod_nsubj(docWhole, docreq):
    for sentence in docreq.sents:
        root_token = sentence.root
        for child in root_token.children:
            if child.dep_ == "nsubj" : 
               value = get_nounchunk(docWhole, child.text) 
               if value != "NONE" :
                  return root_token.text, value
               return root_token.text, child.text                                             

def who_realized(doc, quoi, dequi ):
    for token in doc:
        if token.head.text == quoi  and dequi in token.text :
            token_head = token.head
            for child in token_head.children:
                if  child.dep_ == "nmod" and child.pos_ == "PROPN": # Trouver le sujet du verbe
                    return  child.text
    return "NONE"  

doc = nlp(text)

for question in question_list:
    doc_requst = nlp(question)
    if forme2_qui (doc_requst):
       quoi, dequi = retrieve_nmod_nsubj(doc, doc_requst)
       print(f' quoi : {quoi} de qui : {dequi} par qui : {who_realized(doc, quoi, dequi) }')  
```

  - **Position des mots interrogatifs** : Ces mots peuvent se placer en début ou en fin de phrase selon le style et l'emphase désirée.

```python
import spacy
import re

# Charger le modèle français
nlp = spacy.load("fr_core_news_sm")

# Liste des mots interrogatifs
interrogative_words = {'comment', 'laquelle', 'lequel', 'lesquelles', 'lesquels', 'où', 'pourquoi', 'quand', 'que', 'qui', 'quoi', 'quel', 'quelle', 'quels', 'quelles'}

def analyser_phrase_interrogative_(phrase,docWhole):
    doc = nlp(phrase)
    elements = {
        'mot_interrogatif': None,
        'verbe': None,
        'racine' : None,
        'sujet': None,
        'complement': None,
        'chunk' : None,
        'point_interrogation': doc[-1].text == "?"
    }

    def get_nounchunk(doc,target_token):
        for chunk in doc.noun_chunks:
            if target_token.lower() in chunk.text.lower():
                return chunk.text
        return target_token  

    for sentence in doc.sents:
        root_token = sentence.root
    #    print(f'root_token {root_token}')
        elements['racine'] = str(root_token)
        for token in doc:
    #        print(f'token {token} {token.head}  {token.dep_}  {token.lemma_}')
            if token.pos_ == "PRON" and token.lemma_.lower() in {'qui', 'que', 'quoi', 'quel', 'quelle', 'quels', 'quelles', 'où', 'comment', 'pourquoi', 'quand'}:
                elements['mot_interrogatif'] = token.lemma_.lower()
            elif token.pos_ == "VERB" and not elements['verbe']:
                elements['verbe'] = token.lemma_
            elif token.dep_ in {'nsubj','nsubj:pass'} and not elements['sujet']:
                elements['sujet'] = token.text 
            elif token.dep_ in ["nmod"] and not elements['complement']:
                elements['complement'] = token.text
                elements['chunk'] = get_nounchunk(docWhole, token.head.text)  
        return elements

doc = nlp(text)    
#question = "La Joconde a été peinte par qui ?"
question="Le Radeau de la Méduse est une peinture réalisée par qui ?"
elements = analyser_phrase_interrogative_(question,doc)  
elements  
```

 - Elle parcourt les éléments, excluant le point d'interrogation, et ajoute les valeurs non nulles à la liste des mots-clés.

```python
def analyser_elements(elements):
    mots_cles = []
    #print(f'elements: {elements}')
    for key, value in elements.items():
        if value and key not in ['mot_interrogatif', 'point_interrogation']:
            mots_cles.append((key,value))
    return mots_cles

mots_cles = analyser_elements(elements)
mots_cles
```

```python
import string

def nettoyer_phrase(phrase):
    # Créer une liste de caractères en excluant la ponctuation, puis la rejoindre
    return ''.join(char for char in phrase if char not in string.punctuation)

def comparer_elements(phrases, elements):   
    correspondances = []
    elements = dict(elements)
    for phrase_ in phrases:
        phrase = nettoyer_phrase(str(phrase_))
        score = 0  # Initialiser le score pour chaque phrase
        words_in_phrase = phrase.lower().split()  # Convertir la phrase en mots et en minuscules
       # print(words_in_phrase)
        
        # Vérifier la correspondance du verbe
        if 'verbe' in elements and elements['verbe'] in words_in_phrase:
            score += 1  # Ajouter 1 au score pour le verbe
        
        # Vérifier la correspondance du sujet
        if 'racine' in elements and elements['racine'].lower() in words_in_phrase:
            score += 1  # Ajouter 1 au score pour le sujet
        
        # Vérifier la correspondance de l'objet
        if 'sujet' in elements and elements['sujet'].lower() in words_in_phrase:
            score += 1  # Ajouter 1 au score pour l'objet
        
        # Vérifier la correspondance des compléments
        if 'complement' in elements :
            for complement in elements['complement']:    
                if complement.lower() in words_in_phrase:
                    score += 1  # Ajouter 1 au score pour chaque complément trouvé
        
        correspondances.append((phrase, score))  # Ajouter la phrase et son score
    return correspondances

def filtrer_phrases_maximum(correspondances):
    # Trouver le score maximum
    max_score = max(score for _, score in correspondances)
    # Filtrer les phrases avec le score maximum
    phrases_max = [phrase for phrase, score in correspondances if score == max_score]
    return phrases_max

doc_corpus = nlp(text_corpus)
sentences_corpus  = [ sent for sent in doc_corpus.sents] 
texte_source = filtrer_phrases_maximum(comparer_elements(sentences_corpus, mots_cles))
print(f'texte_source : {texte_source}')
print(f'mots_cles : {mots_cles}')
```
