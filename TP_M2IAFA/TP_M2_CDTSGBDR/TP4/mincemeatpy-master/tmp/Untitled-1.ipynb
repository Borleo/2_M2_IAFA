```python
import re

def text_to_sentences(text):
    sentence_endings = re.compile(r'(?<=[.!?]) +')
    sentences = sentence_endings.split(text.strip())
    return sentences
```

```python
def load_stop_words(filename):
    with open(filename, 'r') as f:
         return set(word.strip().lower() for word in f)
```

```python
def load_file(filename):
    with open(filename, 'r') as f:
         return f.read()
```

```python
import string
import re

def nettoyer_texte(texte):
    # Supprimer le caractère '\ufeff'
    #print(f'--------- nettoyer_texte : in')
    ma_chaine = texte
    if isinstance(ma_chaine, list) : 
        ma_chaine = ' '.join(texte)
        #print(f'nettoyer_texte ma_chaine : {ma_chaine}')
    ma_chaine = ma_chaine.replace('\ufeff', '')
    ma_chaine = ma_chaine.translate(str.maketrans('', '', string.punctuation))   
    ma_chaine = ma_chaine.lower()
    mots = re.findall(r'\b\w+\b', ma_chaine)
    #print(f'nettoyer_texte : {mots}')
    return [' '.join(mots)]
```

```python
import os

def lire_fichiers(chemin_repertoire):
    contenu_fichiers = []
    
    for nom_fichier in os.listdir(chemin_repertoire):
        chemin_complet = os.path.join(chemin_repertoire, nom_fichier)
        if os.path.isfile(chemin_complet):
            try:
                with open(chemin_complet, 'r', encoding='utf-8') as fichier:
                    contenu = fichier.read()
                    contenu_fichiers.append(contenu)
            except Exception as e:
                print(f"Erreur lors de la lecture de {nom_fichier}: {str(e)}")
                
    return contenu_fichiers
```

```python
from collections import Counter

def process_document(doc_dict):
    doc_num = doc_dict['num']
    text = doc_dict['text']
    words = text.split()
    term_count = Counter(words)
    total_terms = len(words)
    term_frequency = {term: count / total_terms for term, count in term_count.items()}
    
    return {
        'num': doc_num,
        'term_count': term_count,
        'total_terms': total_terms,
        'term_frequency': term_frequency
    }

def process_documents(documents):
    return {f"doc_{doc['num']}": process_document(doc) for doc in documents}

def add_document(new_document, document_data):
    doc_num = new_document['num']
    document_data[f"doc_{doc_num}"] = process_document(new_document)
    return document_data

```

```python
# Liste de documents, chacun étant un dictionnaire
documents = [
    {'num': 1, 'text': "Ceci est le premier document."},
    {'num': 2, 'text': "Ce document est le deuxième document."},
    {'num': 3, 'text': "Et voici le troisième."},
    {'num': 4, 'text': "Est-ce le premier document ?"}
]

# Traitement des documents
document_data = process_documents(documents)

# Affichage des informations pour le premier document
print(document_data['doc_1'])

# Ajout d'un nouveau document
new_doc = {'num': 5, 'text': "Un cinquième document à ajouter."}
document_data = add_document(new_doc, document_data)

# Vérification du nouveau document ajouté
print(document_data['doc_5'])

```
