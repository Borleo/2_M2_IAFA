### `Linear Image Denoising`

`1. Le cadre probabiliste`
 - On part d’une image vraie $x_0 ∈ R^N$ (par exemple l’image propre, en 1D ou 2D rangée en vecteur).
 - On observe une image bruitée y où $y = x_0 + w$, et w est un bruit aléatoire.
 - `Un estimateur de débruitage x est une fonction de y : on écrit x=F(y)`.
 - Comme y dépend du bruit w, l’estimateur x est lui-même un vecteur aléatoire qui varie quand le bruit varie.
 - __Donc :__ 
   - x est un estimateur du débruitage qui ne dépend que de y
   - correspond à une règle qui, à partir de l’image bruitée, produit une image débruitée.

`2. Invariance par translation et convolution`
 - On suppose maintenant que l’estimateur est :
   - linéaire : $F(αy_1+βy_2)=αF(y_1)+βF(y_2)$, invariant par translation
   - si on décale (translate) l’image y, la sortie x est décalée de la même façon.
 - Un résultat classique d’analyse des systèmes dit alors :
   - `Tout opérateur linéaire invariant par translation` 
   - `sur des signaux/images périodiques est une convolution avec un noyau h`.   

 - Considèrons l'estimateur de bruit $x\in\RR^N$ of $x_0$ qui dépend des observations $y$.   
 - Le débruitage par invariance de translation linéaire est une convolution avec un noyau $h$, $ x = x_0 \star h $
 - Où $⋆$ est la convolution périodique. En indices discrets 2D, cela donne : $ (a \star b)_i = \sum_j a(j) b(i-j). $
 - La convolution dans le domaine spatial devient un produit dans le domaine de Fourier
$$ \forall \om, \quad \hat x(\om) = \hat x_0(\om) \hat h(\om). $$
   - $\hat x$, $\hat x_0$,$\hat h$ sont les transformées de Fourier de $x, x_0, h$. 
   - ω est une fréquence (2D pour les images).

```python
cconv = lambda a,b : np.real(np.fft.ifft2(np.fft.fft2(a)*np.fft.fft2(b)))
```

We use here a Gaussian fitler $h$ parameterized by the bandwith $\mu$.
Définit un noyau gaussien 2D normalisé $h_μ$ sur une grille périodique $n×n$, paramétré par la largeur $μ$.

```python
normalize = lambda h : h/sum(h.flatten())
t = np.transpose( np.concatenate( (np.arange(0,n/2), np.arange(-n/2,0) ) ) )
[Y,X] = np.meshgrid(t, t)
h = lambda mu: normalize(np.exp(-(X**2 + Y**2)/ (2*mu**2)))
```

Display the filter $h$ and its Fourier transform.

```python
mu = 10
plt.subplot(1,2, 1)
# np.fft.fftshift(h(mu)) sert à recentrer le noyau gaussien hμ  
# pour qu’il soit cohérent avec la convention de la FFT (zéro au centre).
imageplot(np.fft.fftshift(h(mu)))
plt.title('h')
plt.subplot(1,2, 2)
imageplot(np.fft.fftshift(np.real(np.fft.fft2(h(mu)))))
plt.title('$\hat h$');
```

```python
imageplot(h(mu))
```

Shortcut for the convolution with $h$.

```python
denoise = lambda x,mu : cconv(h(mu), x)
```

Display a denoised signal.

```python
imageplot(denoise(y, mu))
```

## Wiener Filtering

We suppose here that $x_0$ is a realization of a random vector $x_0$,
whose distribution is Gaussian with a stationary covariance $c$,
and we denote $P_{X_0}(\om) = \hat c(\om)$ the power-spectrum of
$x_0$.


Recall that $w$ is a realization of a random vector $W$
distributed according to $\Nn(0,\si^2 \text{Id})$.


The (oracle) optimal filter minimizes the risk
$$ R(h) = \EE_{W,X_0}( \norm{ X_0 - h \star (X_0 + W) }^2 ). $$


One can show that the solution of this problem, the so-called Wiener filter,
is defined as
$$ \forall \om, \quad \hat h(\om) = \frac{ P_{X_0}(\om) }{ P_{X_0}(\om) + \si^2 }. $$


We estimate $ P_{X_0} $ using the periodogram associated to the
realization $x_0$, i.e.
$$ P_{X_0} \approx \frac{1}{N}\abs{\hat x_0}^2. $$

```python
P = 1/N * ( abs(fft2(x0))**2 )
```

Compute the approximate Wiener filter.

```python
h_w = real(ifft2(P / (P + sigma**2)))
```

Note that this is a theoretical filter, because in practice one does not
have access to $x_0$. Display it.

```python
u = fftshift(h_w)
imageplot( u[n/2-10:n/2+10,n/2-10:n/2+10] )
```

Display the denoising result.

```python
imageplot( cconv(y,h_w) )
```

Note that this denoising is not very efficient, because the hypothesis of
stationarity of $X_0$ is not realistic for such piecewise-regular
signal.
