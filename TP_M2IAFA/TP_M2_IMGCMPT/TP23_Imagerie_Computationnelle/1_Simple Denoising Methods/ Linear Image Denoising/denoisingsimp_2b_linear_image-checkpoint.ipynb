### `Linear Image Denoising`

`1. Le cadre probabiliste`
 - On part d’une image vraie $x_0 ∈ R^N$ (par exemple l’image propre, en 1D ou 2D rangée en vecteur).
 - On observe une image bruitée $y$ où $y = x_0 + w$, et $w$ est un bruit aléatoire.
 - `Un estimateur de débruitage x est une fonction de y tel que l'on écrit x=F(y)`.
 - Comme $y$ dépend du bruit $w$, `l’estimateur` $x$ `est un vecteur aléatoire qui varie quand le bruit varie`.
 - _Donc :_ 
   - $x$ est un estimateur du débruitage qui ne dépend que de $y$
   - correspond à une règle qui, à partir de l’image bruitée, produit une image débruitée.

`2. Invariance par translation et convolution`
 - On suppose maintenant que l’estimateur est :
   - linéaire : $F(αy_1+βy_2)=αF(y_1)+βF(y_2)$, invariant par translation
   - si on décale (translate) l’image $y$, la sortie $x$ est décalée de la même façon.
 - Un résultat classique d’analyse des systèmes dit :
 
   - `Tout opérateur linéaire invariant par translation sur des signaux/images périodiques est` 
   - __`une convolution avec un noyau h`__   

Considèrons l'estimateur de bruit $x\in\R^N$, $x_0$ dépend des observations $y$, le débruitage par invariance de translation linéaire est une convolution avec un noyau h.
- $x = x_0 \star h $ où $\large ⋆$ est la convolution périodique. 
- En indices discrets 2D, cela donne : $ (a \star b)_i = \sum_j a(j) b(i-j). $
- La convolution dans le domaine spatial devient un produit dans le domaine de Fourier
$$\large \forall \omega,\quad \hat{x}(\omega) = \hat{x}_0(\omega)\,\hat{h}(\omega)$$
   - $\hat x$, $\hat x_0$, $\hat h$ sont les transformées de Fourier de $x, x_0, h$. 
   - ω est le vecteur de fréquence spatiale, $\hat x_0 (ω)= \hat x_0 (ω_x,ω_y)$
      - $ω = (ω_x,ω_y)$ est un vecteur de fréquences spatiales,
      - $w_x$ mesure à quelle vitesse l’image oscille selon l’axe x,
      - $w_y$ mesure à quelle vitesse l’image oscille selon l’axe y,
- _Une fréquence spatiale élevée_ = des variations très rapides des niveaux de gris (détails fins, contours), 
- _Une fréquence faible_ = variations lentes (zones lisses, structures larges).


`L'outil mathématique “convolution circulaire cconv” entre deux images, devient un estimateur de débruitage linéaire invariant par translation lorsqu'on l’utilise pour filtrer l’image bruitée avec un noyau h, par exemple.`

```python
cconv = lambda a,b : np.real(np.fft.ifft2(np.fft.fft2(a)*np.fft.fft2(b)))
```

`Là`, $\hat x = y ∗ h_μ$ `est l’estimateur` $x$ `linéaire, invariant par translation, réalisé comme convolution (ici périodique) avec le noyau` $h_μ$.

 - `on définit un noyau gaussien 2D normalisé` $h_μ$ 
 - `sur une grille périodique` $n×n$, 
 - `paramétré par la largeur` $μ$.

Cette écriture construit (en 1D) un vecteur d’indices de fréquences “centrées” façon FFT, du type [0,1,2,…,n/2−1,−n/2,…,−2,−1].
 - Dimension de l'image $n \times n$. 
 - $h$ est défini dans le domaine spatial, on construit une grille de coordonnées spatiales discrètes (X,Y), avec meshgrid(tx, ty).
 - On normalize(...) pour que la somme vaille 1 
 - Ce qui nous donne un noyau gaussien 2D, filtre flou spatial (passe‑bas en spatial).

```python
normalize = lambda h : h/sum(h.flatten())
tx = np.transpose( 
                np.concatenate(( 
                np.arange(0,n/2), 
                np.arange(-n/2,0))) 
                )
ty = np.transpose( np.concatenate( (np.arange(0,n/2), np.arange(-n/2,0) ) ) )
[Y,X] = np.meshgrid(tx, ty)
h = lambda mu: normalize(np.exp(-(X**2 + Y**2)/ (2*mu**2)))
```

 - `Affiche le filtre` $h$ `et sa transformée de Fourier`.
 - `np.fft.fftshift(h(mu))` sert à recentrer le noyau gaussien $h_μ$  
 - pour qu’il soit cohérent avec la convention de la FFT (zéro au centre).
 - np.real(...) ne prend que la partie réelle 
   - La TF d’une gaussienne réelle et centrée est réelle et positive en continu, 
   - Les petites parties imaginaires viennent des erreurs numériques.

```python
mu = 10
plt.subplot(1,2, 1)
imageplot(np.fft.fftshift(h(mu)))
plt.title('Noyau h dans le domaine spatial')
plt.subplot(1,2, 2)
imageplot(np.fft.fftshift(np.real(np.fft.fft2(h(mu)))))
plt.title('$\hat h$ transformée de Fourier de h');
```

L’image  montre à h sans fftshift, donc la gaussienne centrée en (0,0), c’est‑à‑dire… dans les coins de l’image.

```python

plt.figure(figsize=(6,3))
imageplot(h(mu))
```

`denoise devient un estimateur de débruitage linéaire invariant par translation lorsqu'on utilise cconv “convolution circulaire” entre deux images, pour filtrer l’image bruitée avec un noyau h`

```python
denoise = lambda x,mu : cconv(h(mu), x)
```

Display a denoised signal.

```python

plt.subplot(1, 2, 1)
imageplot(y)
plt.title("Noisy")

plt.subplot(1, 2, 2)
imageplot(denoise(y, mu))
plt.title(f"Denoised, mu={mu}")

plt.show()
```

- Choisir une liste de μ (par ex. mulist = np.linspace(0.1, 3.5, 31) ou la tienne).
- Pour chaque μ, calculer l’image débruitée x_μ = denoise(y,μ).
- Calculer l’erreur oracle $|| x_μ − x_0 ||$ (on peut prendre la norme Frobenius ou L2).
- Tracer cette erreur en fonction de μ.
- Choisir $μ_{opt} = argmin_μ || x_μ − x_0 ||$, et utiliser cette valeur dans la suite (affichage de l’image débruitée “optimale”).

```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.linalg import norm

mulist = np.linspace(0.01, 3.5, 50)  # ou ta liste [0.0001, 0.001, ...]
errors = np.zeros(len(mulist))

for i, mu in enumerate(mulist):
    x_mu = denoise(y, mu)
    errors[i] = norm(x_mu - x0)  # norme Frobenius par défaut

plt.figure(figsize=(5, 3))
plt.plot(mulist, errors, '-o')
plt.xlabel(r'$\mu$')
plt.ylabel(r'$||x_\mu - x_0||$')
plt.title('Oracle denoising error vs mu')
plt.grid(True)
plt.show()

# mu optimal
i_opt = np.argmin(errors)
mu_opt = mulist[i_opt]
print("mu_opt =", mu_opt)

# image débruitée optimale
x_opt = denoise(y, mu_opt)
imageplot(x_opt)
```

## `Wiener Filtering`

- On considère ici $x_0$ comme une **réalisation** d’un vecteur aléatoire $X_0$, gaussien, de covariance **stationnaire** $c$. 
- On note alors le **spectre de puissance** de $X_0$ par $P_{X_0}(\omega) = \hat c(\omega)$
- Le bruit $w$ est une réalisation d’un vecteur aléatoire $W$ de loi gaussienne centrée, de covariance $\sigma^2 \mathrm{Id}$, c’est‑à‑dire $W \sim \mathcal{N}\bigl(0,\sigma^2 \mathrm{Id}\bigr)$
- On cherche un filtre linéaire invariant $h$ qui **minimise le risque quadratique moyen** $R(h) = \mathbb{E}_{W,X_0}\!\left( \bigl\| X_0 - h \star (X_0 + W) \bigr\|^2 \right)$
- On peut montrer que le minimiseur de ce problème est le **filtre de Wiener**, donné en domaine de Fourier par  $\forall \omega,\quad \hat h(\omega) = \frac{P_{X_0}(\omega)}{P_{X_0}(\omega) + \sigma^2}$
- En pratique, on ne connaît pas $P_{X_0}$, on l’**estime** via le périodogramme de la réalisation $x_0$ : $P_{X_0}(\omega) \approx \frac{1}{N}\,|\hat x_0(\omega)|^2$

```python
P = 1/N * ( abs(np.fft.fft2(x0))**2 )
```

Compute the approximate Wiener filter.

```python
h_w = np.real(np.fft.ifft2(P / (P + sigma**2)))
```

Note that this is a theoretical filter, because in practice one does not
have access to $x_0$. Display it.

```python
u = np.fft.fftshift(h_w)
n = u.shape[0]  # ou u.shape[1] si carré, assumez n défini ainsi
center = n // 2
imageplot(u[center-10:center+10, center-10:center+10])
```

Display the denoising result.

```python
imageplot( cconv(y,h_w) )
```

Note that this denoising is not very efficient, because the hypothesis of stationarity of $X_0$ is not realistic for such piecewise-regular signal.
