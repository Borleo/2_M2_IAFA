### `Calcul du produit de convolution complet dans le domaine spatial`.

Le “produit de convolution complet dans le domaine spatial” veut dire : calculer la **convolution linéaire** classique entre une image et un noyau, sans tronquer les bords, donc avec un résultat de taille agrandie.

### 1. Formule 2D discrète (mode *full*)

Soit une image $I$ de taille $M\times N$ et un noyau $K$ de taille $m\times n$.
La convolution 2D discrète complète est

$$
(I * K)(p,q) = \sum_{i=0}^{M-1}\sum_{j=0}^{N-1} I(i,j)\,K(p-i,\,q-j),
$$

pour tous les indices $p,q$ où les termes existent (en pratique, on considère $I(i,j)=0$ en dehors de l’image → **padding zéro**).
Le résultat a alors une taille

$$
(M + m - 1)\times(N + n - 1),
$$

qu’on appelle parfois convolution “full” (par opposition à *same* ou *valid*).

Intuitivement, on fait “glisser” le noyau partout, même quand il ne recouvre qu’une partie de l’image (complété par des zéros), et on garde **tous** les pixels calculés, y compris aux bords.

Tu veux que je te fasse un petit exemple numérique complet (image 3×3, noyau 2×2) avec tous les calculs pour illustrer la taille $4\times 4$ en mode full, ou plutôt un rappel des modes *valid/same/full* en tableau ?



$$
H = \begin{bmatrix}
1 & 4 & 7\\
2 & 5 & 8\\
3 & 6 & 9
\end{bmatrix}.
$$

### 2. Choisissons un noyau 2×2

Par exemple

$$
K = \begin{bmatrix}
1 & 0\\
0 & 1
\end{bmatrix}.
$$

On rappelle que la **convolution** tourne le noyau à 180° (pour ce noyau symétrique, ça ne change rien) et on utilise un padding zéro autour de $H$.

### 3. Taille du résultat en mode *full*

Pour une image $3\times 3$ et un noyau $2\times 2$, la convolution complète a taille

$$
(3+2-1)\times(3+2-1) = 4\times 4.
$$

### 4. Calcul explicite pixel par pixel

On pose le résultat $Y = H * K$, de taille 4×4, indices $p,q \in \{0,1,2,3\}$.
Formule :

$$
Y(p,q) = \sum_{i=0}^{2}\sum_{j=0}^{2} H(i,j)\,K(p-i,\,q-j),
$$

avec $H(i,j)=0$ si $(i,j)$ sort du 3×3, et $K(u,v)=0$ si $(u,v)$ sort du 2×2.

On calcule :

- $Y(0,0)$ : seul $i=0,j=0$ donne des indices valides dans $K$ (0,0).
$\Rightarrow Y(0,0)=H(0,0)K(0,0)=1\cdot1=1.$
- $Y(0,1)$ : contributions possibles
    - $i=0,j=1$ → $K(0,0)$ → $4\cdot1=4$.
    - $i=0,j=0$ → $K(0,1)$ (mais $K(0,1)=0$).
Donc $Y(0,1)=4.$

En continuant de la même façon pour tous les $p,q$, on obtient finalement

$$
Y = H*K =
\begin{bmatrix}
1 & 4 & 7 & 0\\
2 & 6 & 12 & 8\\
3 & 12 & 18 & 9\\
0 & 3 & 6 & 9
\end{bmatrix}.
$$

Chaque pixel de $Y$ est une somme de produits $H(i,j)\,K(\cdot,\cdot)$, en tenant compte du noyau qui “déborde” sur des zéros autour de l’image.

Si tu veux, je peux refaire le même calcul avec un noyau plus “classique” (par ex. un petit filtre moyenneur ou Sobel) pour que tu voies mieux l’effet sur le contenu de H.
